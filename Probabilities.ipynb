**Sum rule**

p (a) = p (a, b) + p (a,∼ b) 
∼ means not b
So this states that the probability of a being true, p (a), is equal to the
joint probability that both a and b are true, p (a, b), plus the joint
probability that a is true and b is false, p (a,∼ b). If the variable b
can take on several values (i.e., more than just true and false), then
calculating p (a) requires summing p (a, b) for each possible value
of b. Note that p (a) is often referred to as a *marginal probability*
in this context when calculated by summing over all the values of
another variable in a joint probability distribution

**Product Rule**

p(a|b)p(b) = p(a, b) = p(b|a)p(a)

p (a, b|c) = p (a|b, c) p (b|c)

**Baye's Theorem**

p (b|a) = p (a|b) * p(b) / p (a)

in logarithmic form multiplications become + and divisions - 

ln p (a, b) = ln p (b|a) + ln p (a)
ln p (b|a) = ln p (a, b) − ln p (a)



```python
# raining = a
# cloudy = b 

p_cloudy = 0.7 
p_not_cloudy = 0.3 

""" Here we assume that we already know conditional probabilities"""

# probability of raining
p_raining_given_cloudy = 0.6  # P(a | b)
p_raining_given_not_cloudy = 0.01 # P(a | ∼b )

#prob of not raining
p_not_raining_given_cloudy = 0.4  # P(∼a | b)
p_not_raining_given_not_cloudy = 0.99 # P(∼a | ∼b)

""" Let's calculate joint probabilities """

# Here we basically mupltiply each prior probability for each conditional probability
# P(a , b) = P(a | b) * P(b)
# notice that if we had p(b | a) we could have done p(a, b) ) p(b|a)*p(a)
# p(a|b)p(b) = p(a, b) = p(b|a)p(a)

p_raining_and_cloudy = p_raining_given_cloudy * p_cloudy 
print(f'p_raining_and_cloudy: {p_raining_and_cloudy}')

p_not_raining_and_cloudy = p_not_raining_given_cloudy * p_cloudy 
print(f'p_not_raining_and_cloudy: {p_not_raining_and_cloudy}')

p_raining_and_not_cloudy = p_raining_given_not_cloudy * p_not_cloudy
print(f'p_raining_and_not_cloudy: {p_raining_and_not_cloudy}')

p_not_raining_and_not_cloudy = p_not_raining_given_not_cloudy * p_not_cloudy 
print(f'p_not_raining_and_not_cloudy: {p_not_raining_and_not_cloudy}')

""" marginal probabilities """

# P(a) = P(a, b) + P(a,∼ b) 

p_raining = p_raining_and_cloudy + p_raining_and_not_cloudy 
print(f'p_raining: {p_raining}')

p_not_raining = p_not_raining_and_cloudy + p_not_raining_and_not_cloudy
print(f'p_not_raining: {p_not_raining}')

""" Posterior probabilities calculated with Logaritimic """

# ln p (b|a) = ln p (a, b) − ln p (a)

p_cloudy_given_raining = p_raining_and_cloudy / p_raining 
print(f'p_cloudy_given_raining: {p_cloudy_given_raining}')

p_cloudy_given_not_raining = p_not_raining_and_cloudy / p_not_raining 
print(f'p_cloudy_given_not_raining: {p_cloudy_given_not_raining}')

p_not_cloudy_given_raining = p_raining_and_not_cloudy / p_raining
print(f'p_not_cloudy_given_raining: {p_not_cloudy_given_raining}')

p_not_cloudy_given_not_raining = p_not_raining_and_not_cloudy / p_not_raining
print(f'p_not_cloudy_given_not_raining: {p_not_cloudy_given_not_raining}')

""" Show that this rule is true p(a|b)p(b) = p(a, b) = p(b|a)p(a) """
# before we caulculated P(a , b) = P(a | b) * P(b) 
# now that we have p(b|a) we can show that p(a, b) = p(b|a)p(a)

p_raining_and_cloudy_1 = p_cloudy_given_raining * p_raining
print(f'p_raining_and_cloudy_1: {p_raining_and_cloudy_1} is equal to p_raining_and_cloudy: {p_raining_and_cloudy}')

p_raining_and_not_cloudy_1 = p_not_cloudy_given_raining * p_raining
print(f'p_raining_and_not_cloudy_1 {p_raining_and_not_cloudy_1} is equal to p_raining_and_not_cloudy: {p_raining_and_not_cloudy}')

p_not_raining_and_cloudy_1 = p_cloudy_given_not_raining * p_not_raining
print(f'p_not_rainig_and_cloudy_1: {p_not_raining_and_cloudy_1} is equal to p_not_raining_and_cloudy: {p_not_raining_and_cloudy}')

p_not_raining_and_not_cloudy_1 = p_not_cloudy_given_not_raining * p_not_raining
print(f'p_not_raining_and_not_cloudy_1: {p_not_raining_and_not_cloudy_1} is equal to p_not_raining_and_not_cloudy: {p_not_raining_and_not_cloudy}')
```
